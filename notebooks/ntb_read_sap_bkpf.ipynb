```python
# Cria uma view com todos os campos da tabela
spark.sql("""
CREATE OR REPLACE VIEW cat_poc_dev.db_sap_s4.vw_bkpf AS
SELECT
  mandt as campo_1,
  bukrs,
  belnr,
  gjahr,
  blart,
  bldat,
  budat,
  waers,
  xblnr,
  usnam,
  tcode,
  cpudt,
  cputm,
  awkey,
  created_at
FROM cat_poc_dev.db_sap_s4.sap_bkpf
""")

# Seleciona os 10 primeiros registros da view criada
df = spark.sql("""
SELECT
  campo_1,
  bukrs,
  belnr,
  gjahr,
  blart,
  bldat,
  budat,
  waers,
  xblnr,
  usnam,
  tcode,
  cpudt,
  cputm,
  awkey,
  created_at
FROM cat_poc_dev.db_sap_s4.vw_bkpf
LIMIT 10
""")

display(df)
```
